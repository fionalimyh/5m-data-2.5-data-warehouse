### Question 1


Let's revisit our `austin_bikeshare_demo` dbt project. Modify the `dim_station.sql` model to include the following columns:

- `avg_duration` (Average of trip durations for trips starting from each station in seconds.)

Then, rebuild the models with the following command to see if the changes are correct:

```bash
dbt run
```


**Answer:**

Paste the `dim_station.sql` model here:

```sql
-- Enter Solution here

WITH stations AS ( -- creates CTE = stations, shows the base station dim records
    SELECT *
    FROM {{ source('austin_bikeshare', 'bikeshare_stations') }}
    WHERE station_id IS NOT NULL
), 

trip_stats AS ( -- creates CTE = trip_stats
    SELECT
        start_station_id,
        AVG(duration_minutes) *60  AS avg_duration   -- calculates the average duration for each start station
    FROM {{ source('austin_bikeshare', 'bikeshare_trips') }}
    WHERE start_station_id IS NOT NULL
        AND duration_minutes IS NOT NULL
        AND duration_minutes > 0
    GROUP BY start_station_id
),

final AS (
    SELECT
        s.station_id,
        s.name              AS station_name,
        s.address,
        s.city_asset_number,
        s.location,
        s.status,
        s.modified_date,
        t.avg_duration
    FROM stations s
    LEFT JOIN trip_stats t -- keeps all station from stations
        ON s.station_id = t.start_station_id
)

SELECT * FROM final -- returns dim model
```
